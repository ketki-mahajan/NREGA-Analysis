
## 1. Project Setup and Library Imports


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("NREGA Python Project Started!")
print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```


## 2. Load the NREGA Dataset


```python
from pathlib import Path

file_path = Path("NREGA Data.csv")

df = pd.read_csv(file_path, sep=";")

print("NREGA dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())
```


## 3. Dataset Overview and Column Names


```python
print("Column names:")
for column in df.columns:
    print(column)
```


## 4. Data Quality Checks


```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values[missing_values > 0])
```

```python
duplicate_count = df.duplicated().sum()
print("Total duplicate rows:", duplicate_count)
```

```python
display(df.describe().T)
```


## 5. Data Cleaning


```python
df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

# Select text columns compatible with your Pandas version
text_columns = df.select_dtypes(include=["object"]).columns

# Remove leading and trailing spaces
for column in text_columns:
    df[column] = df[column].str.strip()

print("Data cleaning completed successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```


## 6. Numeric Data Conversion


```python
numeric_columns = [
    "Total No. of JobCards issued",
    "Total No. of Workers",
    "Total No. of Active Job Cards",
    "Total No. of Active Workers",
    "Approved Labour Budget",
    "Persondays of Central Liability so far",
    "SC persondays",
    "ST persondays",
    "Women Persondays",
    "Average days of employment provided per Household",
    "Average Wage rate per day per person(Rs.)",
    "Total No of HHs completed 100 Days of Wage Employment",
    "Total Households Worked",
    "Total Individuals Worked",
    "Differently abled persons worked",
    "Number of GPs with NIL exp",
    "Total No. of Works Takenup (New+Spill Over)",
    "Number of Ongoing Works",
    "Number of Completed Works",
    "Total Exp(Rs. in Lakhs.)",
    "Wages(Rs. In Lakhs)",
    "Material and skilled Wages(Rs. In Lakhs)",
    "Total Adm Expenditure (Rs. in Lakhs.)"
]

for column in numeric_columns:
    if column in df.columns:
        cleaned = (
            df[column].astype("string")
            .str.strip()
            .str.replace(",", "", regex=False)
        )
        df[column] = pd.to_numeric(cleaned, errors="coerce")

print("Numeric conversion completed.")
display(df.dtypes.to_frame(name="Data Type"))
```


## 7. Overall Worker and Job Card Summary


```python
worker_columns = [
    "Total No. of JobCards issued",
    "Total No. of Workers",
    "Total No. of Active Job Cards",
    "Total No. of Active Workers"
]

overall_workers = df[worker_columns].sum().to_frame(
    name="Total"
)

display(overall_workers)
```


## 8. State-wise Worker Analysis


```python
state_workers = (
    df.groupby("state_name", as_index=False)
    .agg({
        "Total No. of JobCards issued": "sum",
        "Total No. of Workers": "sum",
        "Total No. of Active Job Cards": "sum",
        "Total No. of Active Workers": "sum"
    })
    .sort_values(
        "Total No. of Active Workers",
        ascending=False
    )
)

display(state_workers)
```

```python
top_10 = (
    state_workers
    .head(10)
    .sort_values("Total No. of Active Workers")
)

plt.figure(figsize=(12, 7))

plt.barh(
    top_10["state_name"],
    top_10["Total No. of Active Workers"]
)

plt.title("Top 10 States by Active Workers")
plt.xlabel("Total Active Workers")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```


## 9. Active Worker Percentage by State


```python
state_workers["Active Worker Percentage"] = (
    state_workers["Total No. of Active Workers"]
    .div(
        state_workers["Total No. of Workers"].replace(0, np.nan)
    )
    .mul(100)
    .round(2)
)

display(
    state_workers.sort_values(
        "Active Worker Percentage",
        ascending=False
    )
)
```


## 10. Top 10 Districts by Active Workers


```python

top_10_districts = df[
    [
        "state_name",
        "district_name",
        "Total No. of Active Workers"
    ]
].sort_values(
    by="Total No. of Active Workers",
    ascending=False
).head(10)

display(top_10_districts)
```

```python

plt.figure(figsize=(12, 6))

plt.barh(
    top_10_districts["district_name"][::-1],
    top_10_districts["Total No. of Active Workers"][::-1]
)

plt.title("Top 10 Districts by Active Workers")
plt.xlabel("Number of Active Workers")
plt.ylabel("District")
plt.tight_layout()
plt.show()
```


## 11. Employment Days and Wage Analysis


```python

employment_wage = df[
    [
        "state_name",
        "district_name",
        "Average days of employment provided per Household",
        "Average Wage rate per day per person(Rs.)"
    ]
].copy()

employment_wage = employment_wage.sort_values(
    by="Average days of employment provided per Household",
    ascending=False
)

display(employment_wage.head(10))
```

```python

top_10_employment = employment_wage.head(10)

plt.figure(figsize=(12, 6))

plt.barh(
    top_10_employment["district_name"][::-1],
    top_10_employment[
        "Average days of employment provided per Household"
    ][::-1]
)

plt.title("Top 10 Districts by Average Employment Days")
plt.xlabel("Average Days of Employment per Household")
plt.ylabel("District")
plt.tight_layout()
plt.show()
```


## 12. Budget and Expenditure Analysis


```python

budget_columns = [
    "state_name",
    "district_name",
    "Approved Labour Budget",
    "Total Exp(Rs. in Lakhs.)",
    "Wages(Rs. In Lakhs)",
    "Material and skilled Wages(Rs. In Lakhs)",
    "Total Adm Expenditure (Rs. in Lakhs.)"
]

budget_analysis = df[budget_columns].copy()

print("Budget and Expenditure Analysis")
display(budget_analysis.head(10))
```

```python

budget_analysis["Expenditure Percentage"] = (
    budget_analysis["Total Exp(Rs. in Lakhs.)"]
    / budget_analysis["Approved Labour Budget"].replace(0, np.nan)
) * 100

budget_analysis["Expenditure Percentage"] = (
    budget_analysis["Expenditure Percentage"].round(2)
)

print("Expenditure Percentage Calculated!")
display(
    budget_analysis[
        [
            "state_name",
            "district_name",
            "Approved Labour Budget",
            "Total Exp(Rs. in Lakhs.)",
            "Expenditure Percentage"
        ]
    ].head(10)
)
```


## 13. Work Completion Analysis


```python

work_columns = [
    "state_name",
    "district_name",
    "Total No. of Works Takenup (New+Spill Over)",
    "Number of Ongoing Works",
    "Number of Completed Works"
]

work_analysis = df[work_columns].copy()

print("Work Completion Analysis")
display(work_analysis.head(10))
```

```python

work_analysis["Work Completion Percentage"] = (
    work_analysis["Number of Completed Works"]
    / work_analysis["Total No. of Works Takenup (New+Spill Over)"].replace(0, np.nan)
) * 100

work_analysis["Work Completion Percentage"] = (
    work_analysis["Work Completion Percentage"].round(2)
)

display(
    work_analysis[
        [
            "state_name",
            "district_name",
            "Total No. of Works Takenup (New+Spill Over)",
            "Number of Completed Works",
            "Work Completion Percentage"
        ]
    ].head(10)
)
```


## 14. Women, SC and ST Participation Analysis


```python
participation_columns = [
    "state_name",
    "district_name",
    "Women Persondays",
    "SC persondays",
    "ST persondays"
]

participation_analysis = df[participation_columns].copy()

print("Women, SC and ST Participation Analysis")
display(participation_analysis.head(10))
```

```python
print("Participation Analysis")

# Keep Women, SC and ST persondays separate
participation_columns = [
    "Women Persondays",
    "SC persondays",
    "ST persondays"
]

display(participation_analysis[participation_columns].head(10))
```


# 15. Top 10 Districts by Women Person-Days

```python
top_10_women = df[
    ["state_name", "district_name", "Women Persondays"]
].sort_values(
    by="Women Persondays",
    ascending=False
).head(10)

display(top_10_women)
```

```python
plt.figure(figsize=(12, 6))

plt.barh(
    top_10_women["district_name"][::-1],
    top_10_women["Women Persondays"][::-1]
)

plt.title("Top 10 Districts by Women Person-Days")
plt.xlabel("Women Person-Days")
plt.ylabel("District")
plt.tight_layout()
plt.show()
```


## 16. State-wise NREGA Summary


```python
state_summary = df.groupby("state_name").agg(
    Total_Workers=("Total No. of Workers", "sum"),
    Active_Workers=("Total No. of Active Workers", "sum"),
    Total_Job_Cards=("Total No. of JobCards issued", "sum"),
    Total_Completed_Works=("Number of Completed Works", "sum"),
    Total_Women_Persondays=("Women Persondays", "sum")
).reset_index()

print("State-wise NREGA Summary")
display(state_summary.head(10))
```


## 17. Top 10 States by Active Workers


```python
top_10_states = state_summary.sort_values(
    by="Active_Workers",
    ascending=False
).head(10)

display(top_10_states[["state_name", "Active_Workers"]])
```

```python
plt.figure(figsize=(12, 6))

plt.barh(
    top_10_states["state_name"][::-1],
    top_10_states["Active_Workers"][::-1]
)

plt.title("Top 10 States by Active Workers")
plt.xlabel("Total Active Workers")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```


## 18. Export Analysis Results


```python
from pathlib import Path

output_folder = Path("output")
output_folder.mkdir(exist_ok=True)

state_summary.to_csv(
    output_folder / "state_wise_summary.csv",
    index=False
)

print("State-wise summary exported successfully!")
```

```python
top_10_districts.to_csv(
    output_folder / "top_10_districts_active_workers.csv",
    index=False
)

employment_wage.to_csv(
    output_folder / "employment_and_wages.csv",
    index=False
)

budget_analysis.to_csv(
    output_folder / "budget_and_expenditure.csv",
    index=False
)

work_analysis.to_csv(
    output_folder / "work_completion_analysis.csv",
    index=False
)

participation_analysis.to_csv(
    output_folder / "women_sc_st_participation.csv",
    index=False
)

top_10_states.to_csv(
    output_folder / "top_10_states_active_workers.csv",
    index=False
)

print("All analysis files exported successfully!")
```
